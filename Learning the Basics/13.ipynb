### Pytorch tensors & NumPy
--------------------------------------------------------------------------------------------------

NumPy is a popular scientific Python numerical computing library.

And because of this, PyTorch has functionality to interact with it.

* Data in NumPy, want in PyTorch tensor -> torch.from_numpy(ndarray)
* PyTorch tensor -> NumPy -> torch.Tensor.numpy()

```python
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array).type(torch.float32)
array, tensor
```

```python
array.dtype, tensor.dtype
```

```python
torch.arange(1.0,8.0).dtype
```

```python
array = array + 1
array, tensor
# Note :- While using numpy, if you change an array, then the tensor(which is the same version of that NumPy array) will not change
```

```python
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
numpy_tensor.dtype
# Note :-
# default dtype for torch.tensor -> float 64
# default dtype for numpy.array -> float32
```

```python
tensor = tensor + 1 # Writing tensor += 1 will change the numpy_tensor
tensor, numpy_tensor
```

##### Why 'tensor = tensor + 1' != 'tensor += 1'?
<pre>
When you write:
    tensor = tensor + 1
It creates a new variable (it uses new memory)

But when you write:
    tensor += 1
It just changes the existing memory (instead of creating a new variable).
</pre>

##### Proof:



```python
print(id(tensor))
tensor = tensor + 1
print(f"{id(tensor)} -> different")  # different

tensor = torch.ones(7)
print(id(tensor))
tensor += 1
print(f"{id(tensor)} -> Same as above")  # same
```
