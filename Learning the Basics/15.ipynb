##### About CUDA:
CUDA (Compute Unified Device Architecture) is a platform + a toolkit that allows you to use GPUs for computation(e.g. matrix multiplication), instead of just graphics in games.

This platform is made by NVIDIA, so it only works on NVIDIA GPUs.

--------------------------------------------------------------------------------------------------
##### Why GPUs?
Because GPUs are a lot faster than CPUs. While CPUs can do one task at a time, GPUs can do multiple tasks in parallel, thus reducing the training time dramatically.

GPUs are costly. But we can get free (but limited) access to GPUs online for computation.
