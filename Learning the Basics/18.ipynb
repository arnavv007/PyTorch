## 1. Data (preparing and loading):

```python
# Creating some known data using linear regression formula:
weight = 0.7
bias = 0.3

# Create
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias

X[:10], y[:10], len(X), len(y)
```

#### About the Linear Regression formula:
`y = mx + b`
<pre>
Where:
y = dependent variable
m = Slope of the line
x = independent variable
b = bias
</pre>
This formula creates a line as you can see in the image:<br><br>
<img src = "images/desmos.png" width = "600">
<br>

##### **What will we do?**
1. We will get the data as points on the graph
2. We will create a line (Let's say parallel to the x-axis initially)
3. We will slowly change the slope of the line and use bias to either lift it up or down, so that it best represents the points.
4. Now the line is created and it predicts future points (approximation)
