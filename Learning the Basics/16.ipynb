### 3. Putting tensors (and models) on the GPU
The reason we want out tensors/models on the GPU is because using a GPU results in faster computations.

```python
# Create a tensor (default on the CPU)
tensor = torch.tensor([1,2,3], device="cpu")

#Tensor on the GPU
print(tensor, tensor.device)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
tensor_on_gpu = tensor.to(device)
tensor_on_gpu, device
```

### 4. Moving Tensors back to CPU

```python
# If you want to use NumPy for your tensors, then you'll have to shift to CPU, because NumPy can't leverage with GPU...
tensor_on_gpu.numpy() # -> It will throw an error if you are using CUDA
```

```python
tensor_back_on_CPU = tensor_on_gpu.cpu().numpy()
tensor_back_on_CPU
```

##### Why we can't use NumPy on GPU?
This is because NumPy was made in simpler and older times while GPUs were made in a new era. 
<br><br>
🧠 The real reason

NumPy is designed to:

* Run on CPU only
* Use optimized CPU libraries (BLAS, LAPACK)
* Treat memory as regular RAM

GPUs? Completely different beast:

* Separate memory (VRAM)
* Different architecture (thousands of tiny cores)
* Needs special parallel code (CUDA, etc.)

Numpy just can't talk to GPUs...
