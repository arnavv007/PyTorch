### Splitting data into training and test sets (one of the most important concepts in machine learning in general)

##### **Three datasets:**
1. training set (~ 60-80 %)
2. validation set (~ 10-20 %)
3. test set (~ 10-20 %)

**Generalization:** The ability of a machine learning model to perform well on data it hasn't seen before.

```python
weight = 0.7
bias = 0.3

start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias
X[: 10], y[:10]
```

```python
# Create a train/test split
train_split = int(0.8 * len(X))
print(train_split)
X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

print(len(X_train), len(y_train), len(X_test), len(y_test))
```

#### What did we do above?
We wanted to split our data into training and testing sets with the ratio of 80 : 20. so we multiplied len of our data sets by 0.8 and got 40. Then we took the first 40 elements of X and y, and stored them in the variables X_train and y_train respectively. At last we took the last 10 elements of our X and y data sets and stored them in the variables X_test and y_test respectively. We will use the training variables to train our model and the test variables to test our models later in this course.
