```python
import torch
```

### Finding the min, max, mean, sum, etc. (Tensor aggregation)

```python
tensor = torch.arange(0,100,10) #.type(torch.float32) -> If you add this, then you won't get any error
tensor
```

```python
# Finding the minimum value
torch.min(tensor), tensor.min()
```

```python
# Finding the max value
torch.max(tensor), tensor.max()
```

```python
# Finding the mean
torch.mean(tensor), tensor.mean() # It will throw an error because of wrong dtype
```

```python
# Calculating mean with the right dtype
print(torch.mean(tensor.type(torch.float32)))
print(tensor.type(torch.float32).mean())
```

```python
# Finding the sum
torch.sum(tensor), tensor.sum()
```

```python
# Calculating the product of a tensor
torch.prod(tensor), tensor.prod()
```

##### Tensor aggregation:
It refers to the process of converting a high dimensional tensor into a single value(or few values).

```python
# When you get multiple values instead of a single one:
x = torch.tensor([
                  [1,2,3],
                  [4,5,6]
                ])
x.sum(), x.sum(dim=0), x.sum(dim=1) # You'll get an error if you try dim = 2 or more.

# Remember:
# In PyTorch, 0th dimension means 1st dimension for us, and 1st dimension means 2nd dimension for us. In the above case, the tensor was of 2nd dimension, and writing dim = 2 is just asking for a 3 dimensional array(which is impossible).
```
