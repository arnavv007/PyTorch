### Rules for Matrix multiplicaiton:
1. The inner dimensions must match:
    * (3,2) @ (3,2) won't work
    * (2,3) @ (3,2) will work
    * (3,2) @ (2,3) will work

2. The resuting matrix has the shape of the **outer dimension**
    * (2,3) @ (3,2) -> (2,2)
    * (3,2) @ (2,3) -> (3,3)


```python
# It will throw an error:
A = torch.rand(3,2)
B = torch.rand(3,2)
torch.matmul(A, B)
```

```python
C = torch.rand(2,3)
print(torch.matmul(C,A).shape)
print(torch.matmul(A,C).shape)
# Order Matters!
```

Question:
Why if the inner dimensions don't match, we encounter with an error?

Answer:
Assume two matrices of dimension - (3,2) each:

A = [            
     a b              
     c d<br>
     e f<br>
     ] 
<br><br>
B = [            
     a b              
     c d<br>
     e f<br>
     ]  

Now if we do the matrix multiplication manually, we will see that we multiply a with a and then b with c but there lefts nothing to multiply with c.                

In short:
No. of columns != The no. of rows

For a matrix multiplicaiton:<br>
No. of columns = The no. of rows
