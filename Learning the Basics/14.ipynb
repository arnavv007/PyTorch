### Reproducibility (trying to take random of random)

In short how a neural network w learns:

`start with random numbers -> tensor operations -> update random number to try and make them better representations of the data -> again -> again -> again...`

To reduce the randomness in neural networks and PyTorch comess from the concept of a **random seed.**

Essentially what the random seed does is "flavour" the randomness.

```python
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible tensors

RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED) 
# Note :- you have to set random_seed 2 times... If you don't, then the two tensors will not be equal. This is because setting manual_seed to some value tells PyTorch that you have to start the random number generator from here, and as you intialize more tensors, they will be different because the random seed gets slightly increased after every intialization. that's why if you set the manual_seed only once and then intialize 2 tensors... Then both of the tensors will be different but they will never change doesn't matter how many times you run the code... They will be the exact same everytime.
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```
