### Squeezing, Unsqueezing, and permuting tensors

```python
x = torch.zeros(1,2,3,4,5,6,7,1)
x.shape
```

```python
x.squeeze().shape
```

```python
x_unsqueeze = x.unsqueeze(dim=3)
x_unsqueeze.shape
```

```python
x_original = torch.rand(size=(224,224,3))
x_original.permute(2,0,1).shape, x_original.permute(1,0,2).shape
```

Function                                          Meaning

.permute()            Returns a view of the original input with its dimensions permuted  
                                             (rearranged) dims
