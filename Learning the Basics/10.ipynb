### Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshape an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor.
* stacking - combine multiple tensors on top of each other (vstack) or side by side
* Squeeze - removes all 1 dimensions from a tensor
* Unsqueeze - add a 1 dimension to a target tensor
* Permute - return a view of the input with dimensions permuted(swapped) in a certain way.

```python
x = torch.arange(1., 11.)
x, x.shape, x.dim()
```

```python
# Add an extra dimension

x_reshaped = x.reshape(5,2)
print(x_reshaped, x_reshaped.shape)

print("\n")

x_reshaped_2 = x.reshape(2,5)
print(x_reshaped_2, x_reshaped_2.shape)

# x_reshaped_3 = x.reshape(1,7)
# print(x_reshaped_3, x_reshaped_3.shape) # -> This will throw error because 1*7 != 10(which is the size of the tensor x)
```

```python
# Changing the view
z = x.view(1,10)
z, z.shape, x
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original tensor)
z[:, 0] = 5
z,x
```

```python
# Stack tensors on top of each other
x_stacked =torch.stack([x,x,x,x], dim=0) # hstack
print(x_stacked, x_stacked.shape)

print("\n")

x_stacked_2 =torch.stack([x,x,x,x], dim=1) # vstack
print(x_stacked_2, x_stacked_2.shape)
```

```python
x = torch.tensor([[1,2,3]]) # 2d tensor
print(torch.hstack((x,x,x,x)), torch.vstack((x,x,x,x)))
print(torch.stack([x,x,x,x], dim=0), torch.stack([x,x,x,x], dim = 1))
```
