```python
import torch
```

### Manipulating Tensors

Tensor operations incude:
* Addition
* Subtraction
* Multiplication (element-wise)
* Division
* Matrix multiplication

```python
# Create a tensor and add 10 to it
tensor = torch.tensor([1,2,3])
tensor + 10
```

```python
# Multiplying tensor by 10
tensor *10
```

```python
# subtract 10
tensor - 10
```

```python
# Try out PyTorch in-built functions
torch.mul(10, tensor)
```

```python
torch.add(tensor, 10)
```

### Matrix Multiplication:

Two main ways of performing multiplication in neural networks and deep learning:

1. Element-wise multiplication
2. Matrix multiplication (dot product)

```python
# Element-wise multiplication
print(f"{tensor} * {tensor} = {tensor * tensor}")
```

```python
# Matrix multiplication
torch.matmul(tensor, tensor)
```

```python
%%time
value = 0
for i in range(len(tensor)):
    value += (tensor[i])**2
print(value)
```

```python
%%time
torch.matmul(tensor,tensor)
```

##### torch.mul() vs  torch.matmul():
Let tensor = torch.tensor([1,2,3])

Then:<br>
torch.mul(tensor, tensor) = tensor([1,4,9]) -> Scalar Operation<br>
torch.matmul(tensor,tensor) = tensor(14) -> Matrix multiplication
